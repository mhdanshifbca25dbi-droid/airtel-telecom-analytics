```python
import sqlite3
import pandas as pd

conn = sqlite3.connect(r"C:\Users\dbila\Desktop\airtel-telecom-analytics\data\raw\airtel.db")
def q(sql):
    return pd.read_sql(sql, conn)
```

```python
q("""
SELECT COUNT(*)
FROM customer_features;
""")
```

```python
q("""
SELECT ROUND(AVG(actual_arpu_6m), 2)
FROM customer_features;
""")
```

```python
q("""
SELECT SUM(complaint_count = 0)
FROM customer_features;
""")
```

```python
q("""
SELECT ab_group, COUNT(*)
FROM customer_features
GROUP BY ab_group;
""")
```

```python
q("""
SELECT connection_type,
       COUNT(*) AS subs,
       ROUND(100.0*COUNT(*)/22000, 2) AS pct
FROM customer_features
GROUP BY connection_type
ORDER BY subs DESC;
""")
```

```python
q("""
SELECT plan_name, plan_type, price_inr,
       COUNT(*) AS subs
FROM customer_features
GROUP BY plan_name, plan_type, price_inr
ORDER BY subs DESC
LIMIT 5;
""")
```

```python
q("""
SELECT circle_name, region, COUNT(*) AS subs
FROM customer_features
GROUP BY circle_name, region
ORDER BY subs DESC
LIMIT 5;
""")
```

```python
q("""
SELECT age_band,
       COUNT(*) AS subs,
       ROUND(100.0*COUNT(*)/22000, 1) AS pct
FROM customer_features
GROUP BY age_band
ORDER BY age_band;
""")
```

```python
q("""
SELECT acquisition_channel,
       COUNT(*) AS subs,
       ROUND(100.0*COUNT(*)/22000, 1) AS pct
FROM customer_features
GROUP BY acquisition_channel
ORDER BY subs DESC;
""")
```

```python
q("""
SELECT device_type, os, COUNT(*) AS subs
FROM customer_features
GROUP BY device_type, os
ORDER BY subs DESC;
""")
```

```python
q("""
SELECT tenure_years,
       COUNT(*) AS subs,
       ROUND(100.0*COUNT(*)/22000, 1) AS pct
FROM customer_features
GROUP BY tenure_years
ORDER BY tenure_years;
""")
```

```python
q("""
SELECT ROUND(AVG(stated_arpu), 2) AS avg_arpu,
       COUNT(*) AS subs
FROM customer_features;
""")
```

```python
q("""
SELECT COUNT(*) AS payments,
       ROUND(SUM(amount_inr)/10000000.0, 2) AS crore_inr
FROM recharges
WHERE recharge_date BETWEEN '2025-01-01' AND '2025-06-30';
""")
```

```python
q("""
SELECT ROUND(AVG(amount_inr), 2) AS avg_payment
FROM recharges;
""")
```
